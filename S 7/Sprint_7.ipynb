**Exercici 1: Selecció de Personal**


1. **Obtenir les dades del candidat**
   La funció `obtenir_dades()` pregunta al candidat les seves dades personals i professionals:

   * nom;
   * cognom;
   * telèfon de contacte;
   * anys d'experiència;
   * puntuació de català;
   * puntuació d'Excel;
   * puntuació de la titulació acadèmica.

2. **Validar les dades**
   La funció `validar_dades()` comprova que totes les dades introduïdes siguin correctes:

   * el nom i el cognom han de contenir només lletres;
   * el telèfon ha de contenir només números;
   * els anys d'experiència han de ser numèrics i no poden ser negatius;
   * el nivell de català ha de ser un número entre 0 i 3;
   * el nivell d'Excel ha de ser un número entre 0 i 3;
   * la titulació acadèmica ha de ser un número entre 0 i 3.

   Les puntuacions de català, Excel i titulació tenen un significat concret segons el valor introduït.

3. **Calcular la puntuació del candidat**
   La funció `calcular_puntuacio()` calcula la puntuació total del candidat aplicant la fórmula:

   **(anys d'experiència × 1,5) + català + Excel + titulació acadèmica**

   El resultat és la puntuació final del candidat.

4. **Classificar el candidat**
   La funció `classificar_candidat()` assigna una categoria segons la puntuació obtinguda:

   * puntuació ≥ 12 → **Prioritat alta**;
   * 8 ≤ puntuació < 12 → **Candidat adequat**;
   * puntuació < 8 → **No prioritari**.

5. **Mostrar el resultat individual**
   La funció `mostrar_individual()` mostra al candidat el seu nom, cognom i la puntuació final obtinguda.

6. **Gestionar el procés de selecció**
   La funció `main_seleccio()` crea una estructura per emmagatzemar els candidats i repeteix el procés fins a obtenir **10 candidats vàlids**.

   Per cada candidat:

   * obté les dades;
   * valida les dades;
   * si són correctes, calcula la puntuació;
   * classifica el candidat;
   * mostra el resultat individual;
   * guarda el candidat.

   Si les dades no són vàlides, informa de l'error i torna a demanar les dades.

7. **Mostrar l'informe final**
   Quan s'han obtingut 10 candidats vàlids, la funció `mostrar_informe()` ordena els candidats segons la seva prioritat i, dins de cada prioritat, segons la puntuació obtinguda.

   L'ordre de prioritat és:
   **Prioritat alta → Candidat adequat → No prioritari**

   Finalment, mostra per a cada candidat el nom, el telèfon i la puntuació.





```python
def obtenir_dades():
    nom_cognom = input("Insereix el teu nom i cognom: ")
    telefon = input("Insereix el teu telèfon: ")
    experiencia = int(input("Insereix els anys d'experiència: "))
    catala = int(input("Insereix el nivell de català: "))
    excel = int(input("Insereix el nivell d'Excel: "))
    titulacio = int(input("Insereix el nivell de titulació: "))

    candidat = {
        "nom_cognom": nom_cognom,
        "telefon": telefon,
        "experiencia": experiencia,
        "catala": catala,
        "excel": excel,
        "titulacio": titulacio
    }

    return candidat

def validar_dades(candidat):
  if not candidat["nom_cognom"].replace(" ", "").isalpha():
    return False
  if not candidat["telefon"].isdigit():
    return False
  if candidat["experiencia"] < 0:
     return False
  if candidat["catala"] < 0 or candidat["catala"] > 3:
    return False
  if candidat["excel"] < 0 or candidat["excel"] > 3:
    return False
  if candidat["titulacio"] < 0 or candidat["titulacio"] > 3:
    return False

  return True

def calcular_puntuacio(candidat):
  puntuacio = (
          candidat["experiencia"] * 1.5 +
          candidat["catala"] +
          candidat["excel"] +
          candidat["titulacio"]
      )

  return puntuacio

def classificar_candidat(puntuacio):
  if puntuacio >= 12:
        return "Prioritat alta"
  elif puntuacio >= 8:
        return "Candidat adequat"
  else:
        return "No prioritari"


def mostrar_individual(candidat, puntuacio):
  print(candidat["nom_cognom"])
  print("Puntuació:", puntuacio)


def mostrar_informe(candidats):
    prioritats = {
        "Prioritat alta": 1,
        "Candidat adequat": 2,
        "No prioritari": 3
    }
    candidats_ordenats = sorted(
        candidats,
        key=lambda candidat: (
            prioritats[candidat["prioritat"]],
            -candidat["puntuacio"]
        )
    )
    for candidat in candidats_ordenats:
        print(
            candidat["nom_cognom"],
            candidat["telefon"],
            candidat["puntuacio"]
        )



def main_seleccio():
    candidats = []

    while len(candidats) < 10:
        candidat = obtenir_dades()

        if validar_dades(candidat):
            puntuacio = calcular_puntuacio(candidat)
            prioritat = classificar_candidat(puntuacio)

            candidat["puntuacio"] = puntuacio
            candidat["prioritat"] = prioritat

            mostrar_individual(candidat, puntuacio)
            candidats.append(candidat)

        else:
            print("Error dades")

    mostrar_informe(candidats)












```

## **Nivell 2**

**Exercici 1: Gestió de risc i priorització de pacients**

- netejar_sintomes(sintomes) → limpia y normaliza los síntomas: minúsculas, elimina espacios y cambia espacios por _.
- calcular_risc(sintomes) → busca el nivel de riesgo de cada síntoma y los suma.
- convertir_pacient(pacient) → convierte el texto de un paciente en un diccionario con nombre, edad, síntomas y riesgo total.
- processar_pacients(pacients) → aplica convertir_pacient() a todos los pacientes.
- ordenar_pacients(pacients) → ordena los pacientes de mayor a menor risc_total.
- main_risc() → coordina todo el proceso: procesa los pacientes, los ordena y devuelve el resultado.

```python
import pandas as pd

pacients = [
    "Maria|45|tos, Fiebre Alta, dolor_pecho",
    "Luis|33|dolor_cabeza,faTiga",
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos",
    "Jordi|52|fatiga , tos",
    "Anna|29|fiebre alta,dolor_cabeza"
]

nivells_risc = {
    "fiebre_alta": 3,
    "dificultad_respirar": 5,
    "dolor_pecho": 4,
    "tos": 1,
    "fatiga": 1,
    "dolor_cabeza": 1
}

def convertir_pacient(pacient):
  parts = pacient.split("|")

  nom = parts[0]
  edat = int(parts[1])
  sintomes = parts[2]

  sintomes = netejar_sintomes(sintomes)

  risc_pacient = calcular_risc(sintomes)

  return {
        "nom": nom,
        "edat": edat,
        "sintomes": sintomes,
        "risc_total": risc_pacient
    }

def netejar_sintomes(sintomes):
  parts = sintomes.split(",")
  sintomes_nets = []
  for sintoma in parts:
    sintoma = sintoma.lower()
    sintoma = sintoma.strip()
    sintoma = sintoma.replace(" ", "_")
    sintomes_nets.append(sintoma)

  return sintomes_nets

def calcular_risc(sintomes):
  risc_total = 0
  for sintoma in sintomes:
    risc_total += nivells_risc[sintoma]

  return risc_total

def processar_pacients(pacients):
  pacients_nets = []
  for pacient in pacients:
    pacient_convertit = convertir_pacient(pacient)
    pacients_nets.append(pacient_convertit)

  return pacients_nets

def ordenar_pacients(pacients):
  pacients_ordenats = sorted(
      pacients,
      key=lambda pacient: pacient["risc_total"],
      reverse = True
  )

  return pacients_ordenats

def main_risc():
  pacients_procesats = processar_pacients(pacients)
  pacients_ordenats = ordenar_pacients(pacients_procesats)

  return pacients_ordenats

resultat = main_risc()

df_pacients = pd.DataFrame(resultat)
print(df_pacients)

```

**Exercici 2: Classificació de clients de segurs**

- Calcular risc dels clients, preu estimat pòlissa i la probabilitat de frau.
- Preu_final = preu_base * (1 + edat/100)
- Objectius
» Classificar cada client utilitzant nivells_seguro (sense if/elif).

» Afegir:
La categoria
El preu base
La probabilitat de frau
» Calcular el preu final de la pòlissa amb la fórmula donada.

» Crear un DataFrame o diccionari amb la informació final.

» Ordenar els clients:
Per probabilitat de frau (descendent).
Per preu final.

- obtenir_nivell_seguro(codi) → Busca el nivel de seguro correspondiente al código del cliente y devuelve sus datos.

- calcular_preu_final(preu_base, edat) → Calcula el precio final del seguro teniendo en cuenta el precio base y la edad del cliente.

- llista_clients(clients) → Recorre todos los clientes, obtiene sus datos del seguro, calcula el precio final y añade esta información a cada cliente.

- ordenar_clients_frau(clients) → Ordena los clientes de mayor a menor según su probabilidad de fraude.

- ordenar_clients_preu(clients) → Ordena los clientes de mayor a menor según el precio final del seguro.

- main_segurs() → Coordina todo el proceso: completa los datos de los clientes, los ordena y crea el DataFrame final.

```python
import pandas as pd


clients = [
    {"nom": "Ana", "edat": 42, "codi": 0},
    {"nom": "Carlos", "edat": 30, "codi": 2},
    {"nom": "Isabel", "edat": 55, "codi": 3},
    {"nom": "Jorge", "edat": 40, "codi": 1},
    {"nom": "Marta", "edat": 28, "codi": 0}
]

nivells_seguro = {
    0: {"categoria": "Baix risc", "preu_base": 120, "fraude": 0.01},
    1: {"categoria": "Risc mitjà", "preu_base": 200, "fraude": 0.05},
    2: {"categoria": "Risc alt", "preu_base": 350, "fraude": 0.15},
    3: {"categoria": "Risc crític", "preu_base": 500, "fraude": 0.30}
}

def obtenir_nivell_seguro(codi):
  return nivells_seguro[codi]

def calcular_preu_final(preu_base,edat):
  return preu_base * (1 + edat/100)

def llista_clients(clients):
  for client in clients:
    nivell_seguro = obtenir_nivell_seguro(client["codi"])
    preu_final = calcular_preu_final(nivell_seguro["preu_base"], client["edat"])

    client["categoria"] = nivell_seguro["categoria"]
    client["preu_base"] = nivell_seguro["preu_base"]
    client["fraude"] = nivell_seguro["fraude"]
    client["preu_final"] = preu_final

  return clients

def ordenar_clients_frau(clients):
    clients_ordenats_frau = sorted(
        clients,
        key=lambda client: client["fraude"],
        reverse = True
    )

    return clients_ordenats_frau

def ordenar_clients_preu(clients):
    clients_ordenats_preu = sorted(
        clients,
        key=lambda client: client["preu_final"],
        reverse = True
    )

    return clients_ordenats_preu

def main_segurs():
    clients_complets = llista_clients(clients)
    clients_per_frau = ordenar_clients_frau(clients_complets)
    clients_per_preu = ordenar_clients_preu(clients_complets)

    df_clients = pd.DataFrame(clients_complets)

    print("Clients ordenats per probabilitat de frau:")
    for client in clients_per_frau:
        print(
            client["nom"],
            client["categoria"],
            client["preu_final"],
            client["fraude"]
        )

    print("Clients ordenats per preu final:")
    for client in clients_per_preu:
        print(
            client["nom"],
            client["categoria"],
            client["preu_final"],
            client["fraude"]
        )

    print(df_clients)


main_segurs()



```

**Exercici 1. Generador de contrasenyes segures**

- construir_caracteres() → crea una lista con los tipos de caracteres permitidos según los parámetros: mayúsculas, minúsculas, números y signos.
- crear_contrasenya() → genera la contraseña:
Obtiene los caracteres permitidos.
Añade al menos uno de cada tipo seleccionado.
Completa la longitud indicada con caracteres aleatorios.
Mezcla todos los caracteres.
Devuelve la contraseña como texto (str).

```python
import random
import string


def construir_caracteres(majuscules, minuscules, numeros, signes):
  caracteres = []
  if majuscules:
    caracteres.append(string.ascii_uppercase)
  if minuscules:
    caracteres.append(string.ascii_lowercase)
  if numeros:
    caracteres.append(string.digits)
  if signes:
    caracteres.append(string.punctuation)

  return caracteres

def crear_contrasenya(
    longitud=12,
    majuscules=True,
    minuscules=True,
    numeros=True,
    signes=True
    ):
  lista_caracteres = construir_caracteres(majuscules, minuscules, numeros, signes)
  contrasenya = []
  if majuscules:
    contrasenya.append(random.choice(string.ascii_uppercase))
  if minuscules:
    contrasenya.append(random.choice(string.ascii_lowercase))
  if numeros:
    contrasenya.append(random.choice(string.digits))
  if signes:
    contrasenya.append(random.choice(string.punctuation))
  caracteres_disponibles = "".join(lista_caracteres)
  while len(contrasenya) < longitud:
    contrasenya.append(random.choice(caracteres_disponibles))
  random.shuffle(contrasenya)

  return "".join(contrasenya)


crear_contrasenya()

```

**Exercici 2: Processament automàtic de dades esportives**

```python
import pandas as pd

def recibir_datos(linea):
  linea = linea.strip()
  partes = linea.split("\t")
  resultado = partes[1].split("-")
  goles_locales = int(resultado[0])
  goles_visitante = int(resultado[1])

  return partes[0], goles_locales, partes[2], goles_visitante

def procesar_partidos(lineas):
  estadisticas = {}

  for linea in lineas:
    datos = recibir_datos(linea)

    equipo_local, goles_locales, equipo_visitante, goles_visitantes = datos

    if equipo_local not in estadisticas:
      estadisticas[equipo_local] = {
        "gols_favor": 0,
        "gols_contra": 0,
        "guanyats": 0,
        "empatats": 0,
        "perduts": 0,
        "punts": 0
    }

    if equipo_visitante not in estadisticas:
      estadisticas[equipo_visitante] = {
        "gols_favor": 0,
        "gols_contra": 0,
        "guanyats": 0,
        "empatats": 0,
        "perduts": 0,
        "punts": 0
    }

    estadisticas[equipo_local]["gols_favor"] += goles_locales
    estadisticas[equipo_local]["gols_contra"] += goles_visitantes
    estadisticas[equipo_visitante]["gols_favor"] += goles_visitantes
    estadisticas[equipo_visitante]["gols_contra"] += goles_locales

    if goles_locales > goles_visitantes:
      estadisticas[equipo_local]["guanyats"] += 1
      estadisticas[equipo_local]["punts"] += 3
      estadisticas[equipo_visitante]["perduts"] += 1
    elif goles_locales == goles_visitantes:
      estadisticas[equipo_local]["empatats"] += 1
      estadisticas[equipo_local]["punts"] += 1
      estadisticas[equipo_visitante]["empatats"] += 1
      estadisticas[equipo_visitante]["punts"] += 1
    else:
      estadisticas[equipo_visitante]["guanyats"] += 1
      estadisticas[equipo_visitante]["punts"] += 3
      estadisticas[equipo_local]["perduts"] += 1

  return estadisticas

def maximos_goles(estadisticas, clave):
    equipo_max = None
    max_valor = 0

    for equipo in estadisticas:
        if estadisticas[equipo][clave] > max_valor:
            max_valor = estadisticas[equipo][clave]
            equipo_max = equipo

    return equipo_max


def ordenar_equipos(estadisticas):
  clasificacion_equipos = sorted(
      estadisticas,
      key=lambda equipo: estadisticas[equipo]["punts"],
      reverse = True
  )

  return clasificacion_equipos

def main():
    with open("historic partits.txt", "r") as archivo:
        lineas = archivo.readlines()

    estadisticas = procesar_partidos(lineas)

    equipo_mas_goleador = maximos_goles(estadisticas, "gols_favor")
    equipo_mas_goleado = maximos_goles(estadisticas, "gols_contra")

    print("Equip més golejador:", equipo_mas_goleador)
    print("Equip més golejat:", equipo_mas_goleado)

    print("\nClassificació final:")
    df = pd.DataFrame.from_dict(estadisticas, orient="index")
    df.index.name = "Equip"
    df = df.sort_values(by="punts", ascending=False)
    print(df)

main()
```
